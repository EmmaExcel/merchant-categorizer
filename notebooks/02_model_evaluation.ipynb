# 02 — Model evaluation

Loads the saved evaluation report for the trained artifact and plots the confusion matrix and confidence distribution. Run `make evaluate` first.

```python
import json
import sys
from pathlib import Path

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))

report_path = ROOT / "artifacts" / "current" / "evaluation" / "evaluation_report.json"
report = json.loads(report_path.read_text(encoding="utf-8"))
print("macro_f1:", report["macro_f1"])
print("weighted_f1:", report["weighted_f1"])
print("top1:", report["top1_accuracy"])
print("top3:", report["top3_accuracy"])
```

```python
import pandas as pd

per_class = pd.DataFrame(report["per_class"]).T
per_class.sort_values("f1")
```

```python
from IPython.display import Image, display

eval_dir = ROOT / "artifacts" / "current" / "evaluation"
display(Image(filename=str(eval_dir / "confusion_matrix.png")))
display(Image(filename=str(eval_dir / "confidence_distribution.png")))
```
