```python
from preprocessing.transaction_cleaner import TransactionCleaner
from privacy.redactor_adapter import RedactorAdapter

cleaner = TransactionCleaner(redactor=RedactorAdapter(use_presidio=False))
for raw in [
    "CARD PAYMENT TESCO STORES 3402 LONDON 14/09/2026",
    "BACS ACME LTD PAYROLL REF 12345678",
    "TFR TO SAVINGS 220912",
    "BACS JOHN SMITH 20-45-67 12345678 RENT SEPTEMBER",
]:
    result = cleaner.clean(raw)
    print(raw)
    print("  ->", result.redacted_description)
    print("  ->", result.cleaned_description)
    print()
```
